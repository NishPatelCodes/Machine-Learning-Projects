```python
import numpy as np
```

```python
class Email_classifer:
    def build_vocab(X):
        vocab = set()
        for mssg in X:
            vocab.update(mssg.lower().split())
        return sorted(list(vocab))
        
    def text_to_feature(X,vocab):
        feature = np.zeros((len(X),len(vocab)))
        for i,email in enumerate(X):
            for word in (email.lower().split()):
                feature[i,vocab.index(word)] = 1
        return feature
```

```python
if __name__ == "__main__":

    X = np.array([
    "free offer now",
    "click for free prize",
    "meeting at noon",
    "let’s have lunch",
    "win free money",
    "project discussion"
    ])

    Y = np.array([1, 1, 0, 0, 1, 0])
    model = Email_classifer
    vocab = model.build_vocab(X)
    feature = model.text_to_feature(X,vocab)
    print(vocab)
    print(feature)
```
